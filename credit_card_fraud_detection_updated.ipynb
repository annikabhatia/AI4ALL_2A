### Analysis and Proposed Strategies for Missing Values

The initial analysis reveals a significant number of columns with missing values. Many columns, especially the `D` and `V` features, have over 80% missing data, with some reaching over 90% (e.g., `D7`, `dist2`, `D13`, `D14`, `D12`, `D6`, `D8`, `D9`).


```python
columns_to_drop = missing_values_percentage[missing_values_percentage > 80].index
transaction_data = transaction_data.drop(columns=columns_to_drop)

print(f"Dropped {len(columns_to_drop)} columns with more than 80% missing values.")
print(f"New shape of transaction_data: {transaction_data.shape}")
```

```python
missing_values_after_drop = transaction_data.isnull().sum()
missing_values_percentage_after_drop = (missing_values_after_drop / len(transaction_data)) * 100

#filter to show only columns with missing values and sort them
missing_values_percentage_after_drop = missing_values_percentage_after_drop[missing_values_percentage_after_drop > 0].sort_values(ascending=False)

print("Columns with Missing Values and their Percentages (after dropping >80% missing):")
print(missing_values_percentage_after_drop.head(50))
```

```python
columns_with_missing = missing_values_percentage_after_drop.index

categorical_cols_with_missing = transaction_data[columns_with_missing].select_dtypes(include=['object']).columns
numerical_cols_with_missing = transaction_data[columns_with_missing].select_dtypes(include=[np.number]).columns

print(f"Categorical columns with missing values ({len(categorical_cols_with_missing)}):\n{list(categorical_cols_with_missing)}")
print(f"\nNumerical columns with missing values ({len(numerical_cols_with_missing)}):\n{list(numerical_cols_with_missing)}")
```

### Set the highest missing threshold to 50%.

Previously, when we set the threshold to 70% and imputed the columns with values, it let to very **skewed** results that we were not able to engineer good features for the rest of the ML pipeline. Therefore, we had to come back and make some edits to our EDA process.

```python
HIGH_MISSING_THRESHOLD = 50  # percent

for col in categorical_cols_with_missing:
    pct_missing = missing_values_percentage_after_drop[col]
    if pct_missing > HIGH_MISSING_THRESHOLD:
        transaction_data[col] = transaction_data[col].fillna('missing')
    else:
        mode_value = transaction_data[col].mode()[0]
        transaction_data[col] = transaction_data[col].fillna(mode_value)

print("Imputed categorical columns (mode for <=50% missing, 'missing' sentinel for >50% missing).")

for col in numerical_cols_with_missing:
    median_value = transaction_data[col].median()
    transaction_data[col] = transaction_data[col].fillna(median_value)

print("Missing values in numerical columns imputed with median.")

remaining_missing = transaction_data.isnull().sum()
remaining_missing = remaining_missing[remaining_missing > 0]

if remaining_missing.empty:
    print("\nNo more missing values in the dataset.")
else:
    print("\nRemaining missing values after imputation:")
    print(remaining_missing.sort_values(ascending=False).head())
```

# Feature Engineering

We are creating new features from existing ones that might be relevant for fraud detection, such as transaction velocity, aggregated features by card/email domain, or time-based features.


Extracting the day of the week and hour of the day from 'TransactionDT' will create new time-based features.



```python
import pandas as pd

transaction_data = transaction_data.copy()

transaction_data['TransactionDT_datetime'] = pd.to_datetime(transaction_data['TransactionDT'], unit='s')
transaction_data['Transaction_day_of_week'] = transaction_data['TransactionDT_datetime'].dt.dayofweek
transaction_data['Transaction_hour_of_day'] = transaction_data['TransactionDT_datetime'].dt.hour

print("Created 'Transaction_day_of_week' and 'Transaction_hour_of_day' features.")
print(transaction_data[['TransactionDT', 'Transaction_day_of_week', 'Transaction_hour_of_day']].head())
```

```python
#sort data by card1 and TransactionDT_datetime for accurate rolling window calculations
transaction_data = transaction_data.sort_values(by=['card1', 'TransactionDT_datetime']).reset_index(drop=True)

#calculate transaction velocity (number of transactions per card in a 24-hour window)
window_size_seconds = 86400

# NOTE: groupby(...).rolling(..., on=...) returns a MultiIndex keyed on (card1, TransactionDT_datetime),
# not the original row index -- so merging back on a stored "original_index" column doesn't work.
# Since transaction_data is already sorted by ['card1', 'TransactionDT_datetime'] -- the same order
# groupby+rolling processes rows in -- the result comes back in the same row order as
# transaction_data, so we can assign it back positionally instead of merging.
velocity_series = (
    transaction_data.groupby('card1', sort=True)
    .rolling(window=f'{window_size_seconds}s', on='TransactionDT_datetime', closed='right')
    ['TransactionID']  # Count transactions
    .count()
)

transaction_data['Transaction_velocity_24hr'] = velocity_series.to_numpy()

print("Created 'Transaction_velocity_24hr' feature.")
print(transaction_data[['TransactionDT', 'card1', 'Transaction_velocity_24hr']].head())
```

### Frequency (count) encoding

`card1`, `card2`, `addr1`, and `P_emaildomain` are identifier-like columns. How often a given
value appears in the dataset is itself a useful signal -- brand new or rarely-seen cards/addresses
tend to carry different risk than very common ones.

```python
freq_cols = ['card1', 'card2', 'addr1', 'P_emaildomain']
for col in freq_cols:
    transaction_data[f'{col}_freq'] = transaction_data[col].map(transaction_data[col].value_counts())

print("Created frequency-encoded features:", [f'{c}_freq' for c in freq_cols])
print(transaction_data[[f'{c}_freq' for c in freq_cols]].describe())
```

### Transaction amount features

`TransactionAmt` is right-skewed, so a log transform helps most models. We also pull out the
"cents" portion (fraud sometimes clusters around round numbers) and measure how much a
transaction deviates from that card's typical spend.

Caveat: `TransactionAmt_card_mean`/`_std` are computed over each card's *entire* history in this
dataframe, including transactions that happen after the current row. That's a mild form of
leakage for a time-ordered fraud problem -- fine for exploration, but worth revisiting with an
expanding/backward-looking window once you get to train/test splitting.

```python
transaction_data['TransactionAmt_log'] = np.log1p(transaction_data['TransactionAmt'])
transaction_data['TransactionAmt_decimal'] = (
    (transaction_data['TransactionAmt'] - transaction_data['TransactionAmt'].astype(int)) * 100
).round().astype(int)

transaction_data['TransactionAmt_card_mean'] = transaction_data.groupby('card1')['TransactionAmt'].transform('mean')
transaction_data['TransactionAmt_card_std'] = transaction_data.groupby('card1')['TransactionAmt'].transform('std')
transaction_data['TransactionAmt_to_card_mean_ratio'] = (
    transaction_data['TransactionAmt'] / transaction_data['TransactionAmt_card_mean']
)
transaction_data['TransactionAmt_zscore_card'] = (
    (transaction_data['TransactionAmt'] - transaction_data['TransactionAmt_card_mean'])
    / transaction_data['TransactionAmt_card_std'].replace(0, np.nan)
)
#cards with only one transaction have no std/ratio to compare against --> fill with 0 (neutral).
transaction_data[['TransactionAmt_zscore_card', 'TransactionAmt_to_card_mean_ratio']] = (
    transaction_data[['TransactionAmt_zscore_card', 'TransactionAmt_to_card_mean_ratio']].fillna(0)
)

print("Created TransactionAmt-based features.")
print(transaction_data[['TransactionAmt', 'TransactionAmt_log', 'TransactionAmt_decimal',
                         'TransactionAmt_zscore_card']].head())
```

### Time since the card's last transaction

A card that suddenly transacts again after a long dormant period, or that fires off many
transactions in rapid succession, is a common fraud pattern.

```python
transaction_data['Time_since_last_transaction'] = (
    transaction_data.groupby('card1')['TransactionDT'].diff()
)
#card's first transaction in the data has no prior transaction to compare to.
transaction_data['Time_since_last_transaction'] = transaction_data['Time_since_last_transaction'].fillna(0)

print("Created 'Time_since_last_transaction' feature.")
print(transaction_data[['card1', 'TransactionDT', 'Time_since_last_transaction']].head(10))
```

### Purchaser vs. recipient email domain match

A **mismatch** between `P_emaildomain` (purchaser) and `R_emaildomain` (recipient) is a **common fraud
signal**.  

This feature will be combined with other features to determine fraudulent transactions, since a mismatch between `P_emaildomain` and `R_emaildomain` could also mean it is **gift purchase**.

Note that `R_emaildomain` was mode-imputed earlier (~77% of rows were originally missing),
so this flag is most informative on rows where `R_emaildomain` wasn't imputed.


```python
if 'P_emaildomain' in transaction_data.columns and 'R_emaildomain' in transaction_data.columns:
    transaction_data['email_domain_match'] = (
        transaction_data['P_emaildomain'] == transaction_data['R_emaildomain']
    ).astype(int)
    print("Created 'email_domain_match' feature.")
    print(transaction_data['email_domain_match'].value_counts())
else:
    missing = [c for c in ['P_emaildomain', 'R_emaildomain'] if c not in transaction_data.columns]
    print(f"Skipping 'email_domain_match': {missing} not present in transaction_data.")
    print("This usually means the column was dropped in the >80%-missing step earlier, "
          "or a cell was re-run out of order against modified state. "
          "Run transaction_data.columns.tolist() to check what survived.")
```

## Neural Networks


```python
import tensorflow as tf
from tensorflow.keras import layers, models, Model

def build_residual_block(x, units, dropout_rate=0.3):
    res = layers.Dense(units)(x)
    res = layers.BatchNormalization()(res)
    res = layers.Activation('relu')(res)
    res = layers.Dropout(dropout_rate)(res)

    res = layers.Dense(units)(res)
    res = layers.BatchNormalization()(res)

    if x.shape[-1] != units:
        x = layers.Dense(units)(x)
        x = layers.BatchNormalization()(x)

    out = layers.add([x, res])
    out = layers.Activation('relu')(out)
    return out

#Model Builder
inputs = layers.Input(shape=(354,))
x = layers.Dense(256, activation='relu')(inputs)
x = layers.BatchNormalization()(x)

#Two Residual Blocks
x = build_residual_block(x, 128, dropout_rate=0.3)
x = build_residual_block(x, 64, dropout_rate=0.2)

outputs = layers.Dense(1, activation='sigmoid')(x)
model = Model(inputs=inputs, outputs=outputs, name="Tabular_ResNet")
```

```python
lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=1e-3,
    decay_steps=10000,
    decay_rate=0.9
)
optimizer = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

# Available directly in TensorFlow Addons / Keras natively:
loss = tf.keras.losses.BinaryFocalCrossentropy(
    gamma=2.0,        # Focusing parameter (higher = focuses more on hard cases)
    alpha=0.25        # Balance factor
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_pr_auc',
        mode='max',
        patience=5,
        restore_best_weights=True
    )
]
```

```python
# Neural Networks

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


# Option A: Scikit-learn helper (Recommended)
classes = np.unique(y_train)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
class_weight_dict = dict(zip(classes, weights))

model.compile(
    optimizer=optimizer,
    loss=loss,
    metrics=[tf.keras.metrics.Precision(), tf.keras.metrics.Recall(), tf.keras.metrics.AUC(curve='PR', name='pr_auc')]
)

# Train single architecture (Adjust epochs & batch_size as desired)
history = model.fit(
    X_train,
    y_train,
    epochs=25,
    batch_size=512,  # Larger batch size helps stabilize BatchNorm with 600k rows
    callbacks=callbacks,
    verbose=1,
    class_weight=class_weight_dict
)
```
